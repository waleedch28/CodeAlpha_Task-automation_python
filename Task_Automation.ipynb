```python
import os
import shutil

# Ask user for source and destination paths
source_folder = input("Enter the path of the source folder: ")
destination_folder = input("Enter the path of the destination folder: ")

# Create destination folder if it doesn't exist
os.makedirs(destination_folder, exist_ok=True)

# Counter for moved files
moved_count = 0

# Loop through files and move .jpg
for file_name in os.listdir(source_folder):
    if file_name.lower().endswith(".jpg"):
        source_path = os.path.join(source_folder, file_name)
        destination_path = os.path.join(destination_folder, file_name)
        shutil.move(source_path, destination_path)
        print(f"Moved: {file_name}")
        moved_count += 1

if moved_count == 0:
    print(" No .jpg files found in the source folder.")
else:
    print(f" Done! {moved_count} .jpg file(s) moved to {destination_folder}")
```
